## **REGRESSION**

## LINEAR REGRESSION

### Objective
Predict continuous **Total Sales (INR)** using a single numerical feature (**Price (INR)**).

### Model Type
Supervised Learning — **Regression**

### Variables
- **Independent Variable (X):** Price (INR)  
- **Dependent Variable (y):** Total Sales (INR)

### Mathematical Formula

y = mX + c

(where *m* is the slope and *c* is the intercept)

### Model Performance
- R² Score: **0.5301**  
- Explains **53.01% of total sales variation**

### Interactive Inference
- Takes **1 user input** (Price in INR)  
- Predicts **Total Sales (INR)**  
- Plots a **line of best fit**


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
df = pd.read_excel('flipkart sales.xlsx')
X = df[['Price (INR)']]
y = df['Total Sales (INR)']
```

```python
X = df[['Price (INR)']]
y = df['Total Sales (INR)']
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
score = r2_score(y, y_pred)
print(f"Simple Linear Regression R² Score: {score:.4f}\n")
```

```python
new_price = float(input("Enter Price (INR): "))
new_data = pd.DataFrame([[new_price]], columns=['Price (INR)'])
```

```python
y_answer = model.predict(new_data)
print("Your predicted Total Sales is:", y_answer[0])
```

```python
plt.scatter(X, y, color='blue', alpha=0.4, label='Actual Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line')
plt.scatter(new_price, y_answer[0], color='green', s=150, zorder=5, label='Your Prediction')

plt.xlabel("Price (INR)")
plt.ylabel("Total Sales (INR)")
plt.title("Simple Linear Regression (Full Dataset)")
plt.legend()
plt.show()
```

## MULTIPLE LINEAR REGRESSION

### Objective
Predict continuous **Total Sales (INR)** using multiple independent features (**Price, Quantity Sold, Discount %**).

### Model Type
Supervised Learning — **Multi-variable Regression**

### Variables
- **Independent Variables (X):**  
  - Price (INR)  
  - Quantity Sold  
  - Discount %  
- **Dependent Variable (y):**  
  - Total Sales (INR)

### Mathematical Formula

y = m<sub>1</sub>X<sub>1</sub> + m<sub>2</sub>X<sub>2</sub> + m<sub>3</sub>X<sub>3</sub> + c 

### Model Performance
- R² Score: **0.8853**  
- Explains **88.53% of total sales variation**

### Key Insight
Adding **Quantity Sold** and **Discount %** features significantly improves accuracy compared to single linear regression.

### Interactive Inference
- Takes **3 user inputs** (Price, Quantity, Discount %)  
- Predicts **Total Sales (INR)**  
- Plots **Actual vs. Predicted values**


```python
X = df[['Price (INR)', 'Quantity Sold', 'Discount %']]
y = df['Total Sales (INR)']
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
score = r2_score(y, y_pred)
print(f"Multiple Linear Regression R² Score: {score:.4f}\n")
```

```python
new_price = float(input("Enter Price (INR): "))
new_qty = float(input("Enter Quantity Sold: "))
new_discount = float(input("Enter Discount %: "))

new_data = pd.DataFrame([[new_price, new_qty, new_discount]], 
                        columns=['Price (INR)', 'Quantity Sold', 'Discount %'])
```

```python
y_answer = model.predict(new_data)
print("Your predicted Total Sales is:", y_answer[0])
```

```python
plt.scatter(y, y_pred, color='blue', alpha=0.4, label='Actual vs Predicted')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', label='Perfect Fit Line')
plt.scatter(y_answer[0], y_answer[0], color='green', s=150, zorder=5, label='Your Prediction Point')

plt.xlabel("Actual Total Sales (INR)")
plt.ylabel("Predicted Total Sales (INR)")
plt.title("Multiple Linear Regression (Full Dataset)")
plt.legend()
plt.show()
```

## LOGISTIC REGRESSION

## Objective
Classify whether an order results in **High Sales (1)** or **Low/Medium Sales (0)** based on transaction features.

## Model Type
Supervised Learning — **LOGISTIC REGRESSION**

## Target Construction
- **High_Sales = 1** if *Total Sales > Median*
- **High_Sales = 0** otherwise

## Variables
- **Independent Variables (X):**
  - Price (INR)
  - Quantity Sold
  - Discount %
- **Dependent Variable (y):**
  - High_Sales (Binary: 0 or 1)

## Mathematical Formula



P(y=1) = 1/ 1+e<sup>-z<sup>



(Sigmoid activation function)

## Model Performance
- R<sup>2</sup> Score: **72.86% (0.7286)** across the dataset

## Interactive Inference
- Takes **3 user inputs** (Price, Quantity, Discount %)
- Predicts the binary class (**0 or 1**)
- Displays probability likelihoods in a **bar chart**


```python
from sklearn.linear_model import LogisticRegression
df['High_Sales'] = (df['Total Sales (INR)'] > df['Total Sales (INR)'].median()).astype(int)
```

```python
X = df[['Price (INR)', 'Quantity Sold', 'Discount %']]
y = df['High_Sales']
model = LogisticRegression()
model.fit(X, y)
```

```python
y_pred_all = model.predict(X)
y_prob_all = model.predict_proba(X)[:, 1]
```

```python
print(f"Model R² Score: {r2_score(y, y_prob_all):.4f}\n")
```

```python
new_price = float(input("Enter Price (INR): "))
new_qty = float(input("Enter Quantity Sold: "))
new_discount = float(input("Enter Discount %: "))

new_data = pd.DataFrame([[new_price, new_qty, new_discount]], 
                        columns=['Price (INR)', 'Quantity Sold', 'Discount %'])
```

```python
y_answer = model.predict(new_data)[0]
probabilities = model.predict_proba(new_data)[0]

print("\nPredicted Class (0 = Low/Med Sales, 1 = High Sales):", y_answer)
print(f"Probability of High Sales: {probabilities[1] * 100:.2f}%")
```

```python
categories = ['Low/Med Sales (0)', 'High Sales (1)']
plt.bar(categories, probabilities, color=['skyblue', 'orange'], width=0.4)
plt.ylabel("Probability")
plt.ylim(0, 1)
plt.title(f"Logistic Regression Output (Prediction: {y_answer})")
plt.show()
```
