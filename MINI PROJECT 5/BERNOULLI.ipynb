```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import BernoulliNB
```

```python
csv_filepath = "poetic_vs_normal_dataset.csv"
df = pd.read_csv(csv_filepath)
```

```python
vectorizer = CountVectorizer(binary=True, stop_words="english")
X = vectorizer.fit_transform(df["text"])
y = df["label"]
```

```python
model = BernoulliNB()
model.fit(X, y)
print("=== Model Trained Successfully on CSV Dataset ===")
print("Classes: [0 = Normal, 1 = Poetic]\n")
```

```python
label_map = {0: "Normal", 1: "Poetic"}

print("Enter sentences or words to classify. Type 'exit' or 'done' to stop and view the map.")
print("-" * 65)

while True:
    user_input = input("\nEnter text: ").strip()
    
    # Exit condition
    if user_input.lower() in ["exit", "done", "quit"]:
        print("\nExiting input mode. Generating heatmap...")
        break
    
    if not user_input:
        print("Please enter valid text.")
        continue
    X_user = vectorizer.transform([user_input])
    prediction = model.predict(X_user)[0]
    probabilities = model.predict_proba(X_user)[0]
    
    predicted_class = label_map[prediction]
    confidence = probabilities[prediction] * 100
    
    # Check if any recognized vocabulary words exist in user input
    feature_names = vectorizer.get_feature_names_out()
    matched_words = [word for word in user_input.lower().split() if word in feature_names]
    
    print(f"--> Prediction  : {predicted_class}")
    print(f"--> Confidence  : {confidence:.2f}%")
    print(f"--> Matched Vocab: {matched_words if matched_words else 'None (All unseen words)'}")
```
